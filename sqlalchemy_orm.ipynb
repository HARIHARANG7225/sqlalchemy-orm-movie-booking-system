# movie_booking_system by using sqlalchemy and orm
## Algorithm
### 1-start of the program
### 2-import required thinks from sqlalchmey.orm and sqlalchemy
### 3-create Base class which inherit DeclrativeBase class 
### 4-create table in pydantic way (class)
### 5-create relationships for created table
### 6-create connection by using create_engine
### 7-using engine create table in database by using Base.meta.create_all()
### 8-execute query by creating session
### 9-End of the program

```python
from sqlalchemy.orm import DeclarativeBase,Mapped,mapped_column,relationship,Session
from sqlalchemy import String,Numeric,Text,ForeignKey,create_engine,select
```

```python
class Base(DeclarativeBase):
    pass
```

```python
class Movies_tb(Base):
    __tablename__ = "movies_table"
    movie_id: Mapped[int] = mapped_column(primary_key=True)
    movie_title: Mapped[str] = mapped_column(String(30))
    movie_duration: Mapped[str] = mapped_column(String(15))

    bookings: Mapped[list["Booking_tb"]] = relationship("Booking_tb", back_populates="movie")


class Customers_tb(Base):
    __tablename__ = "customers_table"
    cust_id: Mapped[int] = mapped_column(primary_key=True)
    cust_name: Mapped[str] = mapped_column(String(20))
    cus_mail: Mapped[str] = mapped_column(String(100))

    bookings: Mapped[list["Booking_tb"]] = relationship("Booking_tb", back_populates="customer")


class Booking_tb(Base):
    __tablename__ = "booking_table"
    bk_id: Mapped[int] = mapped_column(primary_key=True)
    cust_id: Mapped[int] = mapped_column(ForeignKey("customers_table.cust_id"))
    movie_id: Mapped[int] = mapped_column(ForeignKey("movies_table.movie_id"))
    bk_timing: Mapped[str] = mapped_column(String(20))

    customer: Mapped["Customers_tb"] = relationship("Customers_tb", back_populates="bookings")
    movie: Mapped["Movies_tb"] = relationship("Movies_tb", back_populates="bookings")
```

```python
from sqlalchemy import select
with Session(engine) as sess:
    sess.add(mv_1)
    sess.add(mv_2)
    sess.commit()

    stmt = select(Movies_tb)
    select = sess.execute(stmt).scalars().all()
    #print(movie.movie_id, movie.movie_title, movie.movie_duration)
    for select in select:
        print(select.movie_id, select.movie_title, select.movie_duration)

```
